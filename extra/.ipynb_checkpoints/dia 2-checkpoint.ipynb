```python
frutas = ["manzana", "pera", "naranja"]

print(frutas)
print(frutas[0])
print(frutas[1])

for fruta in frutas:
    print(fruta)

frutas.append("plátano")
print(frutas)

frutas.remove("pera")
```

```python
notas = [7, 8, 5, 9, 6]

print("Número de notas:", len(notas))
print("Nota máxima:", max(notas))
print("Nota mínima:", min(notas))
print("Suma:", sum(notas))

media = sum(notas) / len(notas)
print("Media:", media)
```
