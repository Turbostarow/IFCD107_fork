```python
def calculadora():

    a = float(input("Primer número: "))
    operador = input("Operador (+, -, *, /): ")
    b = float(input("Segundo número: "))

    if operador == "+":
        resultado = a + b

    elif operador == "-":
        resultado = a - b

    elif operador == "*":
        resultado = a * b

    elif operador == "/":
        if b != 0:
            resultado = a / b
        else:
            print("No se puede dividir entre cero")
            return

    else:
        print("Operador no válido")
        return

    print("Resultado:", resultado)


calculadora()
```
