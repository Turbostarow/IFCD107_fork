# Demo 04 — K-means

Algoritmo no supervisado que agrupa observaciones alrededor de centroides.

## Objetivos de la demo

- Trabajar sin etiquetas.
- Explorar K mediante la inercia.
- Asignar puntos a clusters.
- Actualizar e interpretar centroides.

## 0. Preparación del entorno

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
np.random.seed(42)
```

## 1. Crear datos sin etiquetas

```python
from sklearn.datasets import make_blobs

X, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.9, random_state=42)

plt.scatter(X[:,0],X[:,1])
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Datos sin etiquetas")
plt.show()
```

## 2. Explorar el número de clusters

```python
from sklearn.cluster import KMeans

inertias=[]
ks=range(1,9)
for k in ks:
    km=KMeans(n_clusters=k,n_init=10,random_state=42)
    km.fit(X)
    inertias.append(km.inertia_)

plt.plot(list(ks),inertias,marker="o")
plt.xlabel("K")
plt.ylabel("Inercia")
plt.title("Método del codo")
plt.show()
```

## 3. Entrenar K-means con K=3

```python
km=KMeans(n_clusters=3,n_init=10,random_state=42)
labels=km.fit_predict(X)

plt.scatter(X[:,0],X[:,1],c=labels)
plt.scatter(km.cluster_centers_[:,0],km.cluster_centers_[:,1],
            marker="X",s=220,edgecolor="black",label="Centroides")
plt.xlabel("X1")
plt.ylabel("X2")
plt.legend()
plt.show()
```

## 4. Inspeccionar los centroides

```python
print("Centroides:")
print(km.cluster_centers_)
print("\nInercia:", km.inertia_)
```
