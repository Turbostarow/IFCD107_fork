# Demo 03 — Regresión Logística

Clasificación binaria mediante una función logística que transforma una combinación lineal en una probabilidad.

## Objetivos de la demo

- Crear datos binarios.
- Entrenar `LogisticRegression`.
- Interpretar probabilidades.
- Visualizar la frontera de decisión.
- Evaluar la clasificación.

## 0. Preparación del entorno

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
np.random.seed(42)
```

## 1. Crear un problema de clasificación binaria

```python
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=250, n_features=2, n_redundant=0,
    n_informative=2, n_clusters_per_class=1,
    class_sep=1.5, random_state=42
)

plt.scatter(X[:,0], X[:,1], c=y)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Datos de clasificación")
plt.show()
```

## 2. Entrenar la regresión logística

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X,y)

print("Coeficientes:", model.coef_)
print("Intercepto:", model.intercept_)
```

## 3. Visualizar la frontera de decisión

```python
xx, yy = np.meshgrid(
    np.linspace(X[:,0].min()-1, X[:,0].max()+1, 250),
    np.linspace(X[:,1].min()-1, X[:,1].max()+1, 250)
)
grid = np.c_[xx.ravel(), yy.ravel()]
prob = model.predict_proba(grid)[:,1].reshape(xx.shape)

plt.contourf(xx,yy,prob,levels=20,alpha=0.35)
plt.contour(xx,yy,prob,levels=[0.5],linewidths=2)
plt.scatter(X[:,0],X[:,1],c=y,edgecolor="k")
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Probabilidad y frontera P(y=1)=0.5")
plt.show()
```

## 4. Evaluar la clasificación

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

pred = model.predict(X)
print("Accuracy :", accuracy_score(y,pred))
print("Precision:", precision_score(y,pred))
print("Recall   :", recall_score(y,pred))
print("F1       :", f1_score(y,pred))
print("\nMatriz de confusión:\n", confusion_matrix(y,pred))
```

## 5. Explorar probabilidades

```python
probs = model.predict_proba(X[:10])[:,1]
pd.DataFrame({"real": y[:10], "P(clase=1)": probs, "pred": pred[:10]})
```
