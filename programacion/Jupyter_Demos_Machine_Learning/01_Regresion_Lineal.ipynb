# Demo 01 — Regresión Lineal

Construcción de una recta de regresión para predecir una variable numérica continua.

## Objetivos de la demo

- Crear datos de ejemplo.
- Ajustar una recta mediante `LinearRegression`.
- Visualizar la recta y los residuos.
- Evaluar con R², MAE y RMSE.
- Realizar una predicción.

## 0. Preparación del entorno

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
np.random.seed(42)
```

## 1. Crear los datos de ejemplo

```python
X = np.array([2,4,5,6,8,9,10], dtype=float)
y = np.array([50,60,65,70,80,85,90], dtype=float)

df = pd.DataFrame({"horas_estudio": X, "nota": y})
display(df)

plt.scatter(X, y)
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.title("Datos de ejemplo")
plt.show()
```

## 2. Entrenar la regresión lineal

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X.reshape(-1, 1), y)

print("Intercepto:", model.intercept_)
print("Pendiente:", model.coef_[0])
```

## 3. Representar la recta ajustada

```python
x_grid = np.linspace(X.min(), X.max(), 100)
y_pred = model.predict(x_grid.reshape(-1, 1))

plt.scatter(X, y, label="Datos reales")
plt.plot(x_grid, y_pred, label="Recta de regresión")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.legend()
plt.show()
```

## 4. Evaluar el modelo

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

pred = model.predict(X.reshape(-1,1))
print("R²   :", r2_score(y, pred))
print("MAE  :", mean_absolute_error(y, pred))
print("RMSE :", mean_squared_error(y, pred)**0.5)
```

## 5. Realizar una predicción

```python
horas = np.array([[7]])
print("Predicción para 7 horas:", model.predict(horas)[0])
```
