# Demo 02 — Regresión Polinómica

Extensión de la regresión lineal para modelar relaciones curvas mediante características polinómicas.

## Objetivos de la demo

- Crear una relación no lineal.
- Comparar modelos lineal y polinómico.
- Estudiar el efecto del grado.
- Relacionar complejidad y sobreajuste.

## 0. Preparación del entorno

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
np.random.seed(42)
```

## 1. Datos con relación no lineal

```python
X = np.array([10,12,14,16,18,20,22], dtype=float)
y = np.array([20,35,50,80,75,60,40], dtype=float)

plt.scatter(X,y)
plt.xlabel("Temperatura (°C)")
plt.ylabel("Rendimiento")
plt.title("Relación no lineal")
plt.show()
```

## 2. Comparar regresión lineal y polinómica

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

x_grid = np.linspace(X.min(), X.max(), 200).reshape(-1,1)

linear = LinearRegression().fit(X.reshape(-1,1), y)
poly2 = make_pipeline(PolynomialFeatures(degree=2), LinearRegression()).fit(X.reshape(-1,1), y)

plt.scatter(X,y,label="Datos")
plt.plot(x_grid, linear.predict(x_grid), label="Lineal")
plt.plot(x_grid, poly2.predict(x_grid), label="Polinómica grado 2")
plt.xlabel("Temperatura")
plt.ylabel("Rendimiento")
plt.legend()
plt.show()
```

## 3. Comparar grados

```python
from sklearn.metrics import r2_score, mean_squared_error

for degree in [1,2,3,5]:
    model = make_pipeline(PolynomialFeatures(degree=degree), LinearRegression())
    model.fit(X.reshape(-1,1), y)
    pred = model.predict(X.reshape(-1,1))
    print(f"Grado {degree}: R²={r2_score(y,pred):.3f}, RMSE={mean_squared_error(y,pred)**0.5:.3f}")
```

## 4. Observar el efecto del grado

```python
plt.scatter(X,y,label="Datos")
for degree in [1,2,5]:
    model = make_pipeline(PolynomialFeatures(degree=degree), LinearRegression()).fit(X.reshape(-1,1), y)
    plt.plot(x_grid, model.predict(x_grid), label=f"Grado {degree}")
plt.xlabel("Temperatura")
plt.ylabel("Rendimiento")
plt.legend()
plt.show()
```
