# Demo 05 — K-Nearest Neighbors

Clasificación basada en la similitud con los K vecinos más próximos.

## Objetivos de la demo

- Comprender la distancia.
- Normalizar variables.
- Visualizar la frontera.
- Estudiar el efecto de K.

## 0. Preparación del entorno

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
np.random.seed(42)
```

## 1. Crear datos de clasificación

```python
from sklearn.datasets import make_classification

X,y=make_classification(
    n_samples=220,n_features=2,n_redundant=0,
    n_informative=2,n_clusters_per_class=1,
    class_sep=1.4,random_state=42
)

plt.scatter(X[:,0],X[:,1],c=y)
plt.xlabel("X1"); plt.ylabel("X2")
plt.title("Datos de entrenamiento")
plt.show()
```

## 2. Normalizar y entrenar K-NN

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.25,stratify=y,random_state=42
)

model=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5))
model.fit(X_train,y_train)
```

## 3. Visualizar la frontera de decisión

```python
xx,yy=np.meshgrid(
    np.linspace(X[:,0].min()-1,X[:,0].max()+1,300),
    np.linspace(X[:,1].min()-1,X[:,1].max()+1,300)
)
grid=np.c_[xx.ravel(),yy.ravel()]
zz=model.predict(grid).reshape(xx.shape)

plt.contourf(xx,yy,zz,alpha=0.25)
plt.scatter(X_train[:,0],X_train[:,1],c=y_train,edgecolor="k",label="Train")
plt.scatter(X_test[:,0],X_test[:,1],c=y_test,marker="x",s=70,label="Test")
plt.legend()
plt.xlabel("X1"); plt.ylabel("X2")
plt.title("Frontera K-NN")
plt.show()
```

## 4. Estudiar el efecto de K

```python
from sklearn.metrics import accuracy_score

for k in [1,3,5,15,30]:
    m=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=k))
    m.fit(X_train,y_train)
    print(f"K={k:2d} -> accuracy test={accuracy_score(y_test,m.predict(X_test)):.3f}")
```
