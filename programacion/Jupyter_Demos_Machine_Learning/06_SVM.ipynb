# Demo 06 — Máquinas de Vectores de Soporte

Comparación entre SVM lineal y SVM con kernel RBF para una frontera no lineal.

## Objetivos de la demo

- Comprender el hiperplano y el margen.
- Observar el papel de los kernels.
- Identificar vectores de soporte.
- Explorar el hiperparámetro C.

## 0. Preparación del entorno

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
np.random.seed(42)
```

## 1. Datos no linealmente separables

```python
from sklearn.datasets import make_circles

X,y=make_circles(n_samples=300,factor=0.45,noise=0.08,random_state=42)
plt.scatter(X[:,0],X[:,1],c=y)
plt.xlabel("X1"); plt.ylabel("X2")
plt.title("Datos con frontera no lineal")
plt.show()
```

## 2. Comparar SVM lineal y RBF

```python
from sklearn.svm import SVC

linear=SVC(kernel="linear",C=1).fit(X,y)
rbf=SVC(kernel="rbf",C=1,gamma="scale").fit(X,y)

print("Accuracy SVM lineal:", linear.score(X,y))
print("Accuracy SVM RBF   :", rbf.score(X,y))
```

## 3. Visualizar la frontera

```python
def plot_boundary(model,title):
    xx,yy=np.meshgrid(
        np.linspace(X[:,0].min()-0.2,X[:,0].max()+0.2,300),
        np.linspace(X[:,1].min()-0.2,X[:,1].max()+0.2,300)
    )
    zz=model.predict(np.c_[xx.ravel(),yy.ravel()]).reshape(xx.shape)
    plt.contourf(xx,yy,zz,alpha=0.25)
    plt.scatter(X[:,0],X[:,1],c=y,edgecolor="k")
    plt.title(title); plt.xlabel("X1"); plt.ylabel("X2")
    plt.show()

plot_boundary(linear,"SVM lineal")
plot_boundary(rbf,"SVM con kernel RBF")
```

## 4. Identificar los vectores de soporte

```python
print("Número de vectores de soporte:", len(rbf.support_))
print("Índices de los primeros:", rbf.support_[:10])
```

## 5. Explorar C

```python
for C in [0.1,1,10,100]:
    m=SVC(kernel="rbf",C=C,gamma="scale").fit(X,y)
    print(f"C={C:5} -> accuracy={m.score(X,y):.3f}")
```
