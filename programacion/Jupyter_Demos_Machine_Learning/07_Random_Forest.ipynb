# Demo 07 — Random Forest

Ensemble de árboles de decisión que combina bootstrap, aleatoriedad de variables y agregación de predicciones.

## Objetivos de la demo

- Entrenar un ensemble de árboles.
- Comprender el efecto del número de árboles.
- Evaluar clasificación.
- Interpretar la importancia de variables.

## 0. Preparación del entorno

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
np.random.seed(42)
```

## 1. Crear datos de clasificación

```python
from sklearn.datasets import make_classification

X,y=make_classification(
    n_samples=500,n_features=5,n_informative=4,
    n_redundant=0,n_clusters_per_class=2,
    class_sep=1.1,random_state=42
)

df=pd.DataFrame(X,columns=[f"X{i}" for i in range(1,6)])
df["y"]=y
display(df.head())
```

## 2. Separar entrenamiento y prueba

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.25,stratify=y,random_state=42
)
```

## 3. Entrenar Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf=RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train,y_train)

print("Accuracy train:",rf.score(X_train,y_train))
print("Accuracy test :",rf.score(X_test,y_test))
```

## 4. Analizar la importancia de variables

```python
importance=pd.Series(rf.feature_importances_,index=[f"X{i}" for i in range(1,6)])
importance=importance.sort_values(ascending=False)

display(importance.to_frame("importance"))

importance.sort_values().plot(kind="barh")
plt.xlabel("Importancia")
plt.title("Importancia de variables")
plt.show()
```

## 5. Evaluación más detallada

```python
from sklearn.metrics import classification_report, confusion_matrix

pred=rf.predict(X_test)
print(classification_report(y_test,pred))
print("Matriz de confusión:")
print(confusion_matrix(y_test,pred))
```

## 6. Observar el efecto del número de árboles

```python
for n in [1,5,10,50,100,200]:
    model=RandomForestClassifier(n_estimators=n,random_state=42,n_jobs=-1)
    model.fit(X_train,y_train)
    print(f"n_estimators={n:3d} -> test accuracy={model.score(X_test,y_test):.3f}")
```
