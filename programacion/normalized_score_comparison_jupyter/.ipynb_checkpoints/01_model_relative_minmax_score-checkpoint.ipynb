# Normalized SOC Performance Scores

This notebook uses the same reproducible SOC model-results dataset to compare
two composite scores.

**Formula A — model-relative min-max:** normalization is performed across the
competing models within each land-use category.

**Formula B — SOC-range normalization:** RMSE, MAE and absolute Bias are divided
by the observed SOC range of the corresponding category.

The original metrics remain visible throughout the analysis.

## A. Model-relative min-max normalization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
df=pd.read_csv("soc_model_results.csv")
display(df.head())
```

For higher-is-better metrics:
\[
N_x=(x-x_{\min})/(x_{\max}-x_{\min})
\]
For lower-is-better metrics:
\[
N_x=(x_{\max}-x)/(x_{\max}-x_{\min})
\]
The composite score is the mean of the five normalized components.

```python
def high(s):
    a,b=s.min(),s.max()
    return pd.Series(1.,index=s.index) if np.isclose(a,b) else (s-a)/(b-a)
def low(s):
    a,b=s.min(),s.max()
    return pd.Series(1.,index=s.index) if np.isclose(a,b) else (b-s)/(b-a)

df["N_R2"]=df.groupby("category")["r2"].transform(high)
df["N_RMSE"]=df.groupby("category")["rmse"].transform(low)
df["N_MAE"]=df.groupby("category")["mae"].transform(low)
df["N_Pearson"]=df.groupby("category")["pearson"].transform(high)
df["N_Bias"]=df.groupby("category")["bias"].transform(lambda s: low(s.abs()))
df["Score_A"]=df[["N_R2","N_RMSE","N_MAE","N_Pearson","N_Bias"]].mean(axis=1)
display(df[["category","model","r2","rmse","mae","pearson","bias","Score_A"]].round(4))
```

## Ranking and plot

```python
df["Rank_A"]=df.groupby("category")["Score_A"].rank(ascending=False,method="min")
display(df[["category","model","Score_A","Rank_A"]].sort_values(["category","Rank_A"]).round(4))
fig,ax=plt.subplots(figsize=(11,6))
sns.barplot(data=df,x="category",y="Score_A",hue="model",ax=ax)
ax.set_ylim(0,1.05); ax.set_title("Formula A: model-relative min-max score")
ax.set_ylabel("Score"); plt.xticks(rotation=15); plt.tight_layout(); plt.show()
```

### Interpretation

Formula A is appropriate when the question is **which model performs best
relative to the other models in the same experiment**. Its important
limitation is that adding or removing a competing model can change the
normalized scores of the existing models.
