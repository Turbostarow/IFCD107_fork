# Direct comparison of the two formulations

Both formulas are applied to exactly the same results.

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
df=pd.read_csv("soc_model_results_1.csv")

def high(s):
    a,b=s.min(),s.max()
    return pd.Series(1.,index=s.index) if np.isclose(a,b) else (s-a)/(b-a)
def low(s):
    a,b=s.min(),s.max()
    return pd.Series(1.,index=s.index) if np.isclose(a,b) else (b-s)/(b-a)

df["A_R2"]=df.groupby("category").r2.transform(high)
df["A_RMSE"]=df.groupby("category").rmse.transform(low)
df["A_MAE"]=df.groupby("category").mae.transform(low)
df["A_P"]=df.groupby("category").pearson.transform(high)
df["A_Bias"]=df.groupby("category").bias.transform(lambda s: low(s.abs()))
df["Score_A"]=df[["A_R2","A_RMSE","A_MAE","A_P","A_Bias"]].mean(axis=1)

df["B_R2"]=np.maximum(0,df.r2)
df["B_RMSE"]=np.maximum(0,1-df.rmse/df.soc_range)
df["B_MAE"]=np.maximum(0,1-df.mae/df.soc_range)
df["B_P"]=(df.pearson+1)/2
df["B_Bias"]=np.maximum(0,1-df.bias.abs()/df.soc_range)
df["Score_B"]=df[["B_R2","B_RMSE","B_MAE","B_P","B_Bias"]].mean(axis=1)

df["Rank_A"]=df.groupby("category").Score_A.rank(ascending=False,method="min")
df["Rank_B"]=df.groupby("category").Score_B.rank(ascending=False,method="min")
df["Rank_change"]=df.Rank_B-df.Rank_A
display(df[["category","model","Score_A","Rank_A","Score_B","Rank_B","Rank_change"]].sort_values(["category","Rank_B"]).round(4))
```

```python
plotdf=df.melt(id_vars=["category","model"],value_vars=["Score_A","Score_B"],var_name="formula",value_name="score")
fig,ax=plt.subplots(figsize=(12,6))
sns.barplot(data=plotdf,x="category",y="score",hue="formula",ax=ax)
ax.set_ylim(0,1.05); ax.set_title("Comparison of the two composite scores")
ax.set_ylabel("Score"); plt.xticks(rotation=15); plt.tight_layout(); plt.show()
```

## Recommendation

Use **Formula A** for relative model competition within a fixed experiment.

Use **Formula B** for comparisons across land-use categories with different
SOC ranges. The SOC-range denominator is an observed-data reference and does
not change when competing models are added or removed.

The five original metrics should still be reported alongside either composite
score.
