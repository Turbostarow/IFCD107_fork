# Normalized SOC Performance Scores

This notebook uses the same reproducible SOC model-results dataset to compare
two composite scores.

**Formula A — model-relative min-max:** normalization is performed across the
competing models within each land-use category.

**Formula B — SOC-range normalization:** RMSE, MAE and absolute Bias are divided
by the observed SOC range of the corresponding category.

The original metrics remain visible throughout the analysis.

## B. SOC-range normalization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
df=pd.read_csv("soc_model_results.csv")
display(df[["category","soc_min","soc_max","soc_range"]].drop_duplicates())
```

For each category \(C\):
\[
R_C=SOC_{\max,C}-SOC_{\min,C}
\]
\[
nRMSE_C=RMSE_C/R_C,\quad nMAE_C=MAE_C/R_C,\quad
nBias_C=|Bias_C|/R_C.
\]
Then:
\[
S_{RMSE}=\max(0,1-nRMSE),\quad
S_{MAE}=\max(0,1-nMAE),\quad
S_{Bias}=\max(0,1-nBias)
\]
\[
S_{R^2}=\max(0,R^2),\qquad S_P=(P+1)/2.
\]

```python
df["nRMSE"]=df.rmse/df.soc_range
df["nMAE"]=df.mae/df.soc_range
df["nBias"]=df.bias.abs()/df.soc_range
df["S_RMSE"]=np.maximum(0,1-df.nRMSE)
df["S_MAE"]=np.maximum(0,1-df.nMAE)
df["S_Bias"]=np.maximum(0,1-df.nBias)
df["S_R2"]=np.maximum(0,df.r2)
df["S_Pearson"]=(df.pearson+1)/2
df["Score_B"]=df[["S_R2","S_RMSE","S_MAE","S_Pearson","S_Bias"]].mean(axis=1)
display(df[["category","model","soc_range","r2","rmse","mae","pearson","bias","Score_B"]].round(4))
```

## Ranking and plot

```python
df["Rank_B"]=df.groupby("category")["Score_B"].rank(ascending=False,method="min")
display(df[["category","model","Score_B","Rank_B"]].sort_values(["category","Rank_B"]).round(4))
fig,ax=plt.subplots(figsize=(11,6))
sns.barplot(data=df,x="category",y="Score_B",hue="model",ax=ax)
ax.set_ylim(0,1.05); ax.set_title("Formula B: SOC-range normalized score")
ax.set_ylabel("Score"); plt.xticks(rotation=15); plt.tight_layout(); plt.show()
```

### Interpretation

Formula B answers **how good the prediction is relative to the observed SOC
variability of each land-use category**. Its normalization is independent of
the number and identity of competing models, which makes it preferable for
cross-category comparison when SOC ranges differ.

## 8. Export complete normalized results

The following cell creates the requested publication-ready CSV schema.


```python
# Export the complete SOC-range normalized result table
feature_origin_map = {
    "Random Forest": "Sentinel-2",
    "Extra Trees": "AlphaEarth",
    "Gradient Boosting": "Sentinel-2 + AlphaEarth",
    "Hist. Gradient Boosting": "Sentinel-2",
    "AdaBoost": "AlphaEarth",
}

output = pd.DataFrame({
    "algorithm": df["model"],
    "Feature Origin": df["model"].map(feature_origin_map),
    "r2": df["r2"],
    "rmse": df["rmse"],
    "mae": df["mae"],
    "pearson": df["pearson"],
    "bias": df["bias"],
    "soc_min": df["soc_min"],
    "soc_max": df["soc_max"],
    "soc_range": df["soc_range"],
    "category": df["category"],
})

# Category-level descriptors
category_n = {
    "Cropland": 120,
    "Forest": 90,
    "Grassland": 100,
    "Shrubland": 80,
    "Wetland": 70,
}
soc_mean = {
    "Cropland": 57.5,
    "Forest": 142.5,
    "Grassland": 92.5,
    "Shrubland": 105.0,
    "Wetland": 190.0,
}

output["category_n"] = output["category"].map(category_n)
output["soc_mean"] = output["category"].map(soc_mean)

valid = output["soc_range"] > 0
output["normalization_status"] = np.where(
    valid, "SOC-range normalized", "NOT NORMALIZED: zero SOC range"
)

output["nRMSE"] = np.where(valid, output["rmse"] / output["soc_range"], np.nan)
output["nMAE"] = np.where(valid, output["mae"] / output["soc_range"], np.nan)
output["nBias"] = np.where(valid, output["bias"].abs() / output["soc_range"], np.nan)

output["S_R2"] = np.maximum(0, output["r2"])
output["S_RMSE"] = np.where(valid, np.maximum(0, 1 - output["nRMSE"]), np.nan)
output["S_MAE"] = np.where(valid, np.maximum(0, 1 - output["nMAE"]), np.nan)
output["S_Pearson"] = (output["pearson"] + 1) / 2
output["S_Bias"] = np.where(valid, np.maximum(0, 1 - output["nBias"]), np.nan)

score_cols = ["S_R2", "S_RMSE", "S_MAE", "S_Pearson", "S_Bias"]
output["Score"] = output[score_cols].mean(axis=1)

output["Global_Rank"] = (
    output["Score"].rank(ascending=False, method="min").astype(int)
)
output["Category_Rank"] = (
    output.groupby("category")["Score"]
    .rank(ascending=False, method="min")
    .astype(int)
)

requested_cols = [
    "algorithm", "Feature Origin", "r2", "rmse", "mae", "pearson", "bias",
    "soc_min", "soc_max", "soc_range", "category_n", "soc_mean",
    "normalization_status", "nRMSE", "nMAE", "nBias",
    "S_R2", "S_RMSE", "S_MAE", "S_Pearson", "S_Bias",
    "Score", "Global_Rank", "Category_Rank"
]

output = output[requested_cols]
output.to_csv(
    "normalized_soc_score_output.csv",
    index=False,
    encoding="utf-8-sig"
)

display(output.round(4))
print("CSV exported as normalized_soc_score_output.csv")
```
