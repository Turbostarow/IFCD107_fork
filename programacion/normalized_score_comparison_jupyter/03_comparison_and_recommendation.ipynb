# Comparison — Option A vs Option B

This notebook compares the two output CSV files generated from the same input.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

A = pd.read_csv("option_A_model_relative_minmax_output.csv")
B = pd.read_csv("option_B_soc_range_output.csv")

keys = ["algorithm", "Feature Origin", "category"]
comparison = A[keys + ["Score","Global_Rank","Category_Rank"]].merge(
    B[keys + ["Score","Global_Rank","Category_Rank"]],
    on=keys, suffixes=("_A","_B")
)
comparison["Score_difference_B_minus_A"] = comparison["Score_B"] - comparison["Score_A"]
comparison["Rank_change_B_minus_A"] = comparison["Category_Rank_B"] - comparison["Category_Rank_A"]

display(comparison.sort_values(["category","Category_Rank_B"]).round(4))
```

```python
plotdf = comparison.melt(
    id_vars=["category","algorithm"],
    value_vars=["Score_A","Score_B"],
    var_name="Option", value_name="Score"
)
fig, ax = plt.subplots(figsize=(12,6))
sns.barplot(data=plotdf, x="category", y="Score", hue="Option", ax=ax)
ax.set_ylim(0,1.05)
ax.set_title("Option A vs Option B")
ax.set_ylabel("Composite score")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
